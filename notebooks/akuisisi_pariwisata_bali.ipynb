```python
import json

# Membaca file GeoJSON yang tersedia
with open('/content/export (2).geojson', 'r', encoding='utf-8') as file:
    data = json.load(file)

# Melihat jenis data
print("Jenis data:", data.get("type"))

# Menghitung jumlah objek
print("Jumlah objek:", len(data.get("features", [])))

# Melihat satu contoh data
if data.get("features"):
    print(data["features"][0])
else:
    print("Tidak ada objek features di dalam GeoJSON ini.")
```

```python

import pandas as pd

features = data["features"]

rows = []

for feature in features:
    properties = feature.get("properties", {})
    geometry = feature.get("geometry", {})

    coordinates = geometry.get("coordinates")

    if coordinates:
        if geometry["type"] == "Point":
            longitude, latitude = coordinates[:2]
        else:
            longitude = coordinates[0]
            latitude = coordinates[1]
    else:
        longitude = None
        latitude = None

    rows.append({
        "id": feature.get("id"),
        "nama": properties.get("name"),
        "kategori": properties.get("tourism"),
        "latitude": latitude,
        "longitude": longitude
    })

df = pd.DataFrame(rows)

print("Jumlah data:", len(df))
display(df.head())
```

```python

df.to_csv(
    "/content/wisata_bali.csv",
    index=False,
    encoding="utf-8-sig"
)

print("File CSV berhasil dibuat!")
```
